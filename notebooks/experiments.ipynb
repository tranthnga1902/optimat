# Thí nghiệm mô hình giá vật liệu

Notebook tùy chọn để chạy thí nghiệm và lấy bảng, biểu đồ cho báo cáo. Chạy bằng Jupyter trên máy (`pip install notebook matplotlib`).
Mã huấn luyện nằm trong `src/optimat/predict/train.py`; notebook chỉ gọi lại, không chép logic.

```python
import sys
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(ROOT / 'src'))
import pandas as pd
from optimat.predict.train import train_and_evaluate
```

```python
DATA = ROOT / 'data/sample/prices_sample.csv'   # thay bằng file dữ liệu giá thật
report = train_and_evaluate(str(DATA), str(ROOT / 'models'))
print(report['rows'], report['test_period'])
if report['contains_synthetic']:
    print('CẢNH BÁO:', report['note'])
```

```python
metrics = pd.DataFrame(report['metrics_on_test']).T.round(3)
metrics
```

```python
ax = metrics['MAPE_%'].sort_values().plot.barh(title='MAPE trên tập kiểm tra (%)')
ax.set_xlabel('MAPE (%)')
```
